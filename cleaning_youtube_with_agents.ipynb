```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
from sklearn.naive_bayes import MultinomialNB
from sklearn.cluster import KMeans
from nltk.tokenize import sent_tokenize
import numpy as np
import nltk

nltk.download('punkt')

# Redefining variables locally to ensure they are in scope
t1 = "what up y'all shoot I want to talk about working overnight..." # text1
t2 = "all right so this video I want to talk about Amazon..." # text2
t3 = "what's up family it's your girl Jen ether..." # text3
t4 = "Yes, sir. You just clicked on this video..." # text4
t5 = "My experience of working at Amazon UK..." # text5
t6 = "that's crazy I knew I wasn't tripping..." # text6
t7 = "A lot bro, I couldn't sleep last night..." # text7
t8 = "So today's video is going to be a little bit different..." # text8
t9 = "Hello YouTube, welcome back to my channel..." # text9

documents = [t1, t2, t3, t4, t5, t6, t7, t8, t9]

# 1. Feature Engineering (TF-IDF)
vectorizer = TfidfVectorizer(stop_words='english')
tfidf_matrix = vectorizer.fit_transform(documents)
print(f'TF-IDF Matrix Shape: {tfidf_matrix.shape}')

# 2. Cosine Similarity between any 3 sentences
sentences = [t1[:50], t1[50:100], t1[100:150]] # Dummy sentences for demo
sent_tfidf = vectorizer.transform(sentences)
sim_matrix = cosine_similarity(sent_tfidf)
print('\nCosine Similarity Matrix (3 Sentences):')
print(sim_matrix)

# 3. Supervised Classification (Naive Bayes)
# Label 1 if 'amazon' in text, 0 otherwise
labels = np.array([1 if 'amazon' in d.lower() else 0 for d in documents])
clf = MultinomialNB()
clf.fit(tfidf_matrix, labels)
print(f'\nClassification Predictions: {clf.predict(tfidf_matrix)}')

# 4. Unsupervised Clustering (K-Means)
kmeans = KMeans(n_clusters=2, random_state=42, n_init=10)
kmeans.fit(tfidf_matrix)
print(f'\nClustering Labels: {kmeans.labels_}')
```
